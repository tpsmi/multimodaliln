# Sample and Split Images into Training Dataset

## Overview

This notebook randomly selects images from a source folder and copies them into training and validation sets. Approximately 90% of the selected images are used for training, and the remaining images are used for validation.

The training set is used to train the model. The validation set is used to evaluate its performance during development.

The notebook creates `train` and `val` subfolders in your destination folder. The original images remain in the source folder.

## Prerequisites

Before running the notebook to sample the images, ensure you have the following:

- Images available in `.png`, `.jpg`, or `.jpeg` format.
- Sufficient storage space for copies of the sampled images.
- Python installed on your computer.
- Jupyter Notebook installed to open and run the code.
- The path to the source folder containing your images.
- A destination folder outside the source folder, preferably empty.
- The number of images you want to sample. This must be a positive whole number and cannot exceed the number of available images.

**Note:** Images from different source subfolders must have unique filenames. The code copies them into shared destination folders, so files with matching names will overwrite each other.

## Before Sampling Images

Perform the following steps before sampling the images:

### Install the Required Libraries

Navigate to the following code cell and click **Shift+Enter** to import the required libraries.

```python
import os
import random
import shutil
from tqdm import tqdm
```

**Note:** The `tqdm` library is optional and required to display the progress bar.

### Define the Image Sampling Function

Navigate to the following code snippet.

```python
def sample_images(source_folder, destination_folder, sample_size):
    # Ensure the destination folders exist
    train_folder = os.path.join(destination_folder, 'train')
    val_folder = os.path.join(destination_folder, 'val')
    os.makedirs(train_folder, exist_ok=True)
    os.makedirs(val_folder, exist_ok=True)
    # List all image files in the source folder and its subfolders
    all_images = []
    for root, _, files in tqdm(os.walk(source_folder), desc="Scanning folders"):
        all_images.extend([os.path.join(root, file) for file in files if file.lower().endswith(('.png', '.jpg', '.jpeg'))])
    # Shuffle the list of images to ensure randomness
    random.shuffle(all_images)
    # Check if the sample size is larger than the number of available images
    if sample_size > len(all_images):
        raise ValueError(f"Sample size ({sample_size}) is larger than the number of available images ({len(all_images)})")
    # Randomly sample the images
    sampled_images = random.sample(all_images, sample_size)

    # Calculate the split
    train_size = int(0.9 * sample_size)
    train_images = sampled_images[:train_size]
    val_images = sampled_images[train_size:]
  # Copy the sampled images to the destination folders
    for img in tqdm(train_images, desc="Copying training images"):
        dst_path = os.path.join(train_folder, os.path.basename(img))
        shutil.copy2(img, dst_path)
    for img in tqdm(val_images, desc="Copying validation images"):
        dst_path = os.path.join(val_folder, os.path.basename(img))
        shutil.copy2(img, dst_path)
    print(f"Copied {len(train_images)} images to the training set")
    print(f"Copied {len(val_images)} images to the validation set")
```

Before running the code, specify the following attributes:

- `source_folder`: The path to the folder containing your images
- `destination_folder`: The path where the notebook must create `train` and `val` subfolders.
- `sample_size`: The total number of images to select across both the sets.

Keep the folder path inside the quotation marks in the following format:

```python
source_folder =’/downloads’
destination_folder = ‘/data images/’
sample_size =1000
```

For a sample size of 1000, the notebook selects 900 training images and 100 validation images.

## Start Sampling

Locate and run the following code snippet:

```python
try:
    sample_images(source_folder, destination_folder, sample_size)
    print("\nImage sampling completed successfully!")
    print(f"Training images are in: {os.path.join(destination_folder, 'train')}")
    print(f"Validation images are in: {os.path.join(destination_folder, 'val')}")
except Exception as e:
    print(f"An error occurred: {str(e)}")
```

The notebook displays progress bars while scanning folders and copying images.

## Check the Sampled Images

1. Wait for the message **Image sampling completed successfully** to appear.

   - `train`: training images.
   - `val`: validation images.
3. Confirm that the file counts match the expected split and open several images to check that they copied correctly.
